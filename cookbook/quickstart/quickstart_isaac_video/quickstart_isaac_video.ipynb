Ask Isaac 0.3 Max a question about a video and get a natural-language answer back. This quickstart configures the Perceptron SDK, downloads a sample video, and runs a single Q&A call.

**Note**

Set the `PERCEPTRON_API_KEY` environment variable (or edit the cell below) before running the call. You can swap the prompt or video (MP4 or WebM) for your own at any time.

## Download the sample video

```python
VIDEO_URL = "https://raw.githubusercontent.com/perceptron-ai-inc/perceptron/main/cookbook/_shared/assets/tutorials/isaac_frame_by_frame/surf.mp4"

!curl -L -so surf.mp4 {VIDEO_URL}
```

## Configure the Perceptron client
Authenticate once and point the SDK at Isaac 0.3 Max.

```python
import os

from perceptron import configure, question, video

api_key = os.getenv("PERCEPTRON_API_KEY", "<your Perceptron API key>")
if not api_key or api_key.startswith("<"):
    raise RuntimeError("Set PERCEPTRON_API_KEY or replace the placeholder in this cell.")

configure(
    provider="perceptron",
    model="isaac-0.3-max",
    api_key=api_key,
)

VIDEO_PATH = "surf.mp4"
```

## Ask a question about the video

```python
QUESTION = "What happens in this video?"

result = question(video(VIDEO_PATH), QUESTION)
print(result.text)
```
