## Configure the Perceptron client
Authenticate once and point the SDK at Isaac 0.3 Max.

```python
import os

from perceptron import configure, question, video

api_key = os.getenv("PERCEPTRON_API_KEY", "<your Perceptron API key>")
if not api_key or api_key.startswith("<"):
    raise RuntimeError("Set PERCEPTRON_API_KEY or replace the placeholder in this cell.")

configure(
    provider="perceptron",
    model="isaac-0.3-max",
    api_key=api_key,
)

VIDEO_URL = "https://raw.githubusercontent.com/perceptron-ai-inc/perceptron/main/cookbook/_shared/assets/tutorials/isaac_frame_by_frame/surf.mp4"
```

## Ask a question about the video

```python
QUESTION = "What happens in this video?"

result = question(video(VIDEO_URL), QUESTION)
print(result.text)
```

## Next steps
