# Get started with Isaac 0.2 — Image

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/perceptron-ai-inc/perceptron/blob/main/cookbook/quickstart/quickstart_isaac_0_2/quickstart_isaac_0_2.ipynb)

Ask Isaac 0.2 a question about an image and get a natural-language answer back. This quickstart configures the Perceptron SDK for the `isaac-0.2-1b` model and runs a single Q&A call against a sample image URL — no local download needed.

Isaac 0.2 is an open-weights image vision-language model. The 1B variant is the fast, edge-deployable option; switch to `isaac-0.2-2b-preview` for the reasoning-enabled 2B model.

![Studio scene sample](https://raw.githubusercontent.com/perceptron-ai-inc/perceptron/main/cookbook/_shared/assets/capabilities/qna/studio_scene.webp)

**Note**

Set the `PERCEPTRON_API_KEY` environment variable (or edit the cell below) before running the call. You can swap the prompt or image (URL or local file path) for your own at any time.

## Configure the Perceptron client
Authenticate once and point the SDK at Isaac 0.2. Set `model="isaac-0.2-2b-preview"` if you want the reasoning-capable 2B variant instead.

```python
import os

from perceptron import configure, image, question

api_key = os.getenv("PERCEPTRON_API_KEY", "<your Perceptron API key>")
if not api_key or api_key.startswith("<"):
    raise RuntimeError("Set PERCEPTRON_API_KEY or replace the placeholder in this cell.")

configure(
    provider="perceptron",
    model="isaac-0.2-1b",
    api_key=api_key,
)

IMAGE_URL = "https://raw.githubusercontent.com/perceptron-ai-inc/perceptron/main/cookbook/_shared/assets/capabilities/qna/studio_scene.webp"
```

## Ask a question about the image

```python
QUESTION = "What is in this image?"

result = question(image(IMAGE_URL), QUESTION)
print(result.text)
```

## Optional: enable reasoning (2B Preview only)

Isaac 0.2 2B Preview supports chain-of-thought reasoning. Switch the model and pass `reasoning=True` to see the model's step-by-step thinking alongside the final answer. The 1B variant does not surface reasoning.

```python
configure(
    provider="perceptron",
    model="isaac-0.2-2b-preview",
    api_key=api_key,
)

result = question(image(IMAGE_URL), QUESTION, reasoning=True)
print("Reasoning trace:", result.reasoning)
print("Answer:", result.text)
```

## Next steps
- Swap `IMAGE_URL` for your own image (URL or local file path — `image()` accepts both).
- Tune `QUESTION` toward inspections, retrieval, or scene description.
- Pass `expects="box"`, `"point"`, or `"polygon"` to receive grounded annotations alongside the answer.
- See the [Image Q&A capability guide](https://docs.perceptron.inc/isaac-0.2/capabilities/image-qa) for grounded responses with bounding-box overlays.
- For self-hosting, download the [isaac-0.2-1b weights](https://huggingface.co/PerceptronAI/Isaac-0.2-1B) or [isaac-0.2-2b-preview weights](https://huggingface.co/PerceptronAI/Isaac-0.2-2B-Preview) from Hugging Face.
