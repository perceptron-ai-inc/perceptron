## Configure the Perceptron client
Authenticate once and point the SDK at Isaac 0.3 Max.

```python
import os

from IPython.display import display
from PIL import Image as PILImage

from perceptron import configure, image, ocr

api_key = os.getenv("PERCEPTRON_API_KEY", "<your Perceptron API key>")
if not api_key or api_key.startswith("<"):
    raise RuntimeError("Set PERCEPTRON_API_KEY or replace the placeholder in this cell.")

configure(
    provider="perceptron",
    model="isaac-0.3-max",
    api_key=api_key,
)

LABELS = "grocery_labels.webp"
OCR_PROMPT = "Extract all price tags."
```

## Extract labels from a single frame
Preview the grocery case, run OCR with a custom prompt, and inspect the response.

```python
display(PILImage.open(LABELS))
ocr_result = ocr(image(str(LABELS)), prompt=OCR_PROMPT)
print(ocr_result.text)
```
