## Configure the Perceptron client
Authenticate once and point the SDK at Isaac 0.3 Max.

```python
import os

from IPython.display import display
from PIL import Image as PILImage

from perceptron import caption, configure, image

api_key = os.getenv("PERCEPTRON_API_KEY", "<your Perceptron API key>")
if not api_key or api_key.startswith("<"):
    raise RuntimeError("Set PERCEPTRON_API_KEY or replace the placeholder in this cell.")

configure(
    provider="perceptron",
    model="isaac-0.3-max",
    api_key=api_key,
)

SUBURBAN_STREET = "suburban_street.webp"
SOLAR_ARRAY = "solar_array.webp"
```

## Describe a suburban street
The simplest call requests a concise caption and returns plain text.

```python
display(PILImage.open(SUBURBAN_STREET))
street_caption = caption(image(str(SUBURBAN_STREET)), style="concise", expects="text")
print(street_caption.text)
```

## Request a detailed caption
Switch styles to get a richer narrative description with more nuance and context.

```python
display(PILImage.open(SOLAR_ARRAY))
solar_caption = caption(
    image(str(SOLAR_ARRAY)),
    style="detailed",
    expects="text",
)
print(solar_caption.text)
```

## Conclusion & next steps
- Adjust the `style` argument (`concise`, `detailed`) to match your UX.
- Keep `expects="text"` for pure narrative outputs or request grounding via `expects="box"` / `"point"`.
- Pass `reasoning=True` for richer descriptions of complex scenes.
