## Configure the Perceptron client
Authenticate once and load the cereal box asset from the shared capabilities folder.

```python
import os

from IPython.display import Image as IPyImage
from IPython.display import display

from perceptron import configure, image, ocr

api_key = os.getenv("PERCEPTRON_API_KEY", "<your Perceptron API key>")
if not api_key or api_key.startswith("<"):
    raise RuntimeError("Set PERCEPTRON_API_KEY or replace the placeholder in this cell.")

configure(
    provider="perceptron",
    model="isaac-0.2-2b-preview",
    api_key=api_key,
)

CEREAL_BOX = "mini_wheats.jpeg"
OCR_PROMPT = "Extract brand and product name from this package."
```

## Extract labels from a single frame
Preview the cereal box, run OCR with a custom prompt, and inspect the transcribed text.

```python
display(IPyImage(url=IMAGE_URL, width=400))
ocr_result = ocr(image(str(CEREAL_BOX)), prompt=OCR_PROMPT)
print(ocr_result.text)
```
