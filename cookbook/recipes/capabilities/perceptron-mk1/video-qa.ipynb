## Configure the Perceptron client
Authenticate once and point the SDK at Perceptron Mk1.

```python
import os

from perceptron import configure, question, video

api_key = os.getenv("PERCEPTRON_API_KEY", "<your Perceptron API key>")
if not api_key or api_key.startswith("<"):
    raise RuntimeError("Set PERCEPTRON_API_KEY or replace the placeholder in this cell.")

configure(
    provider="perceptron",
    model="perceptron-mk1",
    api_key=api_key,
)

VIDEO_PATH = "robot_assembly.mp4"
```

## Ask a question about the video
Reasoning is enabled so the model can think through the episode before answering.

```python
QUESTION = "Watch this episode and state the overall goal and sub-goals."

result = question(video(VIDEO_PATH), QUESTION, reasoning=True)

print("--- Reasoning ---")
print(result.reasoning or "(none)")
print("\n--- Answer ---")
print(result.text)
```

## Next steps
- Swap `VIDEO_PATH` for your own MP4 or WebM (URL or local file path; max ~60 seconds at 1 fps sampling).
- Tune `QUESTION` toward summarization, compliance checks, or step decomposition.
- Set `reasoning=False` for faster, lighter responses on simpler queries.
